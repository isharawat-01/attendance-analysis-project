# Impact of 75% Attendance Rule on Students

A Data Analysis and Mathematical Study 

### Introduction

The 75% attendance rule is commonly implemented in educational institutions to ensure student discipline and regular participation in classes. However, many students find it difficult to maintain the required attendance due to various factors such as health issues, academic workload, and personal challenges.

This project aims to analyze the impact of the 75% attendance rule on students using survey data. It also explores whether attendance has a direct relationship with academic performance and examines the difficulty of recovering attendance once it drops below the required threshold.

Additionally, a mathematical approach is used to understand why recovering attendance becomes challenging, providing deeper insight into the practicality of the rule.

### Objectives

- To analyze whether maintaining 75% attendance is realistic for students  
- To study the relationship between attendance and academic performance  
- To understand the level of stress caused by the attendance rule  
- To examine the difficulty of recovering attendance using mathematical analysis

```python
import pandas as pd

df = pd.read_csv("../data/75% Attendance Rule - Student Survey.csv")
df.head()
```

```python
df['Your current attendance (%) '].value_counts().plot(kind='bar')
```

### Distribution of student attendence

The bar chart shows the distribution of students across different attendance ranges.

It can be observed that most students fall in the 50–65% attendance category. 
However, a significant number of students also have attendance below 50%, as well as in the 65–75% and above 75% ranges.

A large portion of students are unable to meet the required attendance criteria, which suggests that the 75% rule may be difficult to maintain consistently.

```python
df['Do you feel stressed due to the 75% attendence rule?'].value_counts().plot(kind='bar')
```

### Stress due to 75% attendance rule

The bar chart represents students' responses to whether they feel stressed due to the 75% attendance rule.

It can be clearly observed that the majority of students answered "Yes", indicating that they feel stressed due to this rule. A smaller number of students responded "Sometimes", while only a few students answered "No".

This suggests that the 75% attendance requirement creates noticeable stress among students. The pressure to maintain the required attendance percentage may negatively impact students' mental well-being.

Overall, the data indicates that while the rule aims to ensure discipline, it also contributes significantly to student stress.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Your current attendance (%) ', hue='Your average marks (%)', data=df)
plt.show()
```

### Relationship Between Attendance and Academic Performance

The graph illustrates the relationship between students' attendance levels and their average marks.

It can be observed that students are distributed across all performance categories regardless of their attendance levels. Even students with low attendance (below 50%) are able to achieve high marks (above 80%), while some students with higher attendance fall into average or lower mark ranges.

This indicates that there is no strong direct relationship between attendance and academic performance. While higher attendance may help in maintaining consistency, it does not guarantee higher marks.

Overall, the data suggests that academic performance depends on multiple factors such as self-study, understanding of concepts, and individual learning ability, rather than attendance alone.

```python
df['If your attendance drops below 75%, is it difficult to recover? '].value_counts().plot(kind='bar')
```

### Difficulty in Recovering Attendance Below 75%

The bar chart shows students' responses to whether it is difficult to recover attendance once it drops below 75%.

It is clearly observed that the majority of students answered "Yes", indicating that they find it difficult to recover their attendance once it falls below the required level. Only a small number of students responded "No" or "Not sure".

This suggests that the attendance system becomes challenging once a student falls behind. Recovering attendance requires consistently attending a large number of classes, which may not always be practical for students.

Overall, the data highlights that maintaining attendance is easier than recovering it after it drops below 75%.

### Mathematical Analysis of Attendance Recovery

Attendance percentage is calculated using the formula:

Attendance % = (Attended Classes / Total Classes) × 100

For example:
If a student has attended 60 out of 100 classes, their attendance is:

60%

To reach the required 75% attendance, the student needs to attend additional classes.

Let x be the number of extra classes attended:

(60 + x) / (100 + x) = 0.75

Solving this:

60 + x = 75 + 0.75x  
0.25x = 15  
x = 60  

This means the student must attend 60 more classes continuously to reach 75%.

This clearly shows that once attendance drops, recovering it becomes very difficult, which supports the survey results where most students reported difficulty in recovery.
