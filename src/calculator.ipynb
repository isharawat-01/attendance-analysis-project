## Attendance Recovery Calculator

This notebook calculates how many additional classes a student needs 
to attend in order to reach the required 75% attendance.

It uses a mathematical approach to increment attendance step-by-step 
until the target percentage is achieved.

```python
def attendance_recovery(attended, total, target=0.75):
    """
    Calculates number of extra classes required to reach target attendance.
    """

    required_classes = 0

    while True:
        new_attendance = (attended + required_classes) / (total + required_classes)
        if new_attendance >= target:
            break
        required_classes += 1

    return required_classes


# Test
if __name__ == "__main__":
    attended = int(input("Enter attended classes: "))
    total = int(input("Enter total classes: "))

    result = attendance_recovery(attended, total)

    print(f"You need to attend {result} more classes to reach 75%")
```

```python
import pandas as pd

df = pd.read_csv('../data/75% Attendance Rule - Student Survey.csv')


attendance_map = {
    'Below 50%': 40,
    '50-65%': 57,
    '65-75%': 70,
    'Above 75%': 85
}

marks_map = {
    'Below 50': 40,
    '50-65': 57,
    '65-80': 72,
    'Above 80': 85
}

df['attendance_num'] = df['Your current attendance (%) '].map(attendance_map)
df['marks_num'] = df['Your average marks (%)'].map(marks_map)

print("Correlation:", df['attendance_num'].corr(df['marks_num']))
```
